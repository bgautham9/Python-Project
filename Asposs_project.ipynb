```python
df['name_length'] = df['name'].astype(str).apply(len)

df['name_length'].describe()
```

```python
df['price'].head()
```

```python
df['price'] = (
    df['price']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
df['price'].dtype
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(x=df['price'])

plt.title("Price Outlier Detection")

plt.show()
```

```python
category_price = df.groupby('category')['price'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=category_price.values,
    y=category_price.index
)

plt.title("Average Price by Category")
plt.xlabel("Average Price")
plt.ylabel("Category")

plt.show()
```

```python
top_colors = df['color'].value_counts().head(15)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_colors.values,
    y=top_colors.index
)

plt.title("Most Popular Colors")
plt.xlabel("Count")
plt.ylabel("Color")

plt.show()
```

```python
df['word_count'] = df['name'].astype(str).apply(lambda x: len(x.split()))

df['word_count'].describe()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df['word_count'], bins=20)

plt.title("Product Name Word Count Distribution")

plt.show()
```

```python
df['description_length'] = df['description'].astype(str).apply(len)

df['description_length'].describe()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df['description_length'], bins=30)

plt.title("Description Length Distribution")

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

numeric_df.corr()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```
